## 1. 문자열의 개요
* **문자열의 개념**: 0개 이상의 문자들이 모인 집합으로, 작은따옴표(`'`)나 큰따옴표(`"`)로 감싸 표현합니다.
* **문자열 연결하기**: `+` 연산자를 사용하여 두 문자열을 하나로 합칠 수 있습니다.
* **문자열 반복하기**: `*` 연산자를 사용하여 문자열을 지정한 횟수만큼 반복할 수 있습니다.

```python
a= 'Hi'
b= 'Python'
print(a,b)

c= "Hi" + "Python"
print(c)

d= "Hi" * 3
print(d)
```

```python
#여러줄의 문자열 - ''' 내용 ''' , *** 내용 ***
a = ''' 이름: 홍길동
나이:25'''
print(a)

b = '이름: 홍길동 \n나이:25'
print(b)
```

```python
#escape문자:
#줄바꿈 \n
print('안녕\n파이썬')

#탭 간격띄우기 \t
print('안녕\t파이썬')

#역슬래시 \\
print('안녕\\파이썬')

#작은 따옴표
print('안녕  \'  파이썬')

#큰 따옴표
print("안녕  \"  파이썬")
```

## 2. 문자열 다루기
* **인덱싱(Indexing)**: 문자열의 특정 위치에 있는 문자에 접근하는 방법입니다. 인덱스는 **0번**부터 시작합니다.
* **슬라이싱(Slicing)**: 문자열의 일부분을 추출하는 방법입니다. `[시작:끝]` 형식에서 끝 인덱스 직전까지 추출합니다.
* **문자열 함수 및 메서드**:
    * `len()`: 문자열의 길이를 반환합니다.
    * `upper()` / `lower()`: 대문자/소문자로 변환합니다.
    * `replace()`: 특정 문자를 다른 문자로 치환합니다.
    * `split()`: 특정 구분자를 기준으로 문자열을 분리하여 리스트로 만듭니다.
    * `count()`: 특정 문자의 출현 횟수를 셉니다.
    * `find()` / `index()`: 특정 문자의 위치를 탐색합니다.

```python
#len(문자열): 문자열의 길이
a = 'abc'
for i in range(len(a)):
    print(a[i], end='')

# for 변수 in 문자열:인덱스 없이 문자에 접근
for i in a:
    print(i, end='')
```

```python
#문자 인덱싱
#문자열[시작인덱스:종료인덱스]

a = 'abcde'
print(a[2:4])
print(a[3:])
print(a[:3])
print(a[1:-2])
```

### 모스 부호 번역기
1. **데이터 구조**: 알파벳 리스트(`alphabet`)와 그에 대응하는 모스 부호 리스트(`morse`)를 생성합니다.
2. **문장 → 모스 부호**:
    * 입력받은 문장을 한 글자씩 분리합니다.
    * `alphabet`에서 해당 글자의 인덱스를 찾습니다.
    * 동일한 인덱스를 가진 `morse`의 요소를 연결하여 출력합니다.
3. **모스 부호 → 문장**:
    * 입력받은 모스 부호를 공백 기준으로 `split()`합니다.
    * `morse` 리스트에서 위치를 찾아 해당하는 알파벳을 조합합니다.


```python
alphabet = 'ABCDEFGHIJKLMNOPQRSTUVWXYZ'
morse = ['.-','-...','-.-.','-..','.','..-.','--.',
         '....','..','.---','-.-','.-..','--','-.',
         '---','.--.','--.-','.-.','...','-','..-',
         '...-','.--','-..-','-.--','--..']

# 🔹 알파벳 → 모스부호
s = input('알파벳 문자열: ').upper()   # 소문자 대비

for ch in s:
    if ch in alphabet:
        print(morse[alphabet.index(ch)], end=' ')
print()

# 🔹 모스부호 → 알파벳
m = input('모스 부호 입력 (공백으로 구분): ')
m_list = m.split()

for code in m_list:
    if code in morse:
        print(alphabet[morse.index(code)], end=' ')
```
