# 📘 파이썬 기초: 04. 반복 구조

## 1. 반복 구조의 개념
* **정의**: 특정 명령이나 문장을 반복해서 실행하는 구조(Iteration)를 의미합니다.
* **필요성**: 동일하거나 유사한 작업을 여러 번 수행해야 할 때(예: 난수 1,000개 생성), 코드를 일일이 작성하는 한계를 극복하고 효율적으로 프로그램을 작성하기 위해 사용합니다.
* **종류**: 파이썬의 대표적인 반복문으로는 `for 문`과 `while 문`이 있습니다.

```python
import random

for i in range(5): #0, 1, 2, 3, 4
  print(random.randint(1,100))
```

## 2. for 문
* **특징**: 주로 반복 횟수가 정해져 있거나 범위가 명확할 때 사용합니다.
* **기본 문법**:
    * `for 변수 in 셀 수 있는 것:` 형식을 사용합니다.
    * '셀 수 있는 것' 뒤에는 반드시 **콜론(:)**을 붙여야 합니다.
    * 반복 실행할 문장은 반드시 **들여쓰기(Indentation)**를 해야 합니다.
* **range() 함수 활용**:
    * `range(5)`: 0부터 4까지 5번 반복합니다.
    * `range(1, 11)`: 1부터 10까지 반복합니다.
    * `range(시작값, 종료값, 증가값)`

```python
for i in  range(5):
  print(1,5)
```

### 문제: 전 직원의 월급 총액 구하기

* 직급이 2급이상, 월급이 본봉+상여금(본봉의 30%)

* 직급이 2급 미만, 월급이 본봉+상여금(본봉의 20%)

* 변수 emplovee-num 직원 3명

* 변수 total 합계

* 변수 rank 직급 입력 받고

* 변수 basic_salary 본봉 입력받고

* 변수 bonus 상여금

* 변수 salary 월급



```python
employee_num = 3
total = 0

for i in range(employee_num): # 0, 1, 2
  rank = int(input('직급: '))
  basic_salary = int(input('본봉'))
  if rank >= 2:
    bonus = basic_salary * 0.3
  else:
    bonus = basic_salary * 0.2
  salary = basic_salary + bonus
  total += salary

  print(f'전 직원 월급 총액: {total}')
```

### 적정 심박수 구하기

* 카르보넨 공식
* 묵표 심박수 = (((220-skdl)-안정 시 심박수)*운동 강도)+ 안정 시 심박수
* 최대 운동 강도를 40%, 50%, 60%, 70%, 80%에 해당하는 목표 심박수를 계산하는 식
* 변수 intensity 운동 강도(만약 40%면 40/100)
* 변수 resting_HR 안정 시 심박수
* 변수 target_HR 목표 심박

*나이age 입력 받고
*안정 시 심박수 입력 받고
* 운동 강도 40~80%, 10% 단위로 목표 심박수를 구해야 함.
*강도 몇 %일 때 목표 심박수가 얼마다 라고 출력

```python
age = int(input('나이: '))
resting_HR = int(input('안정 시 심박수: '))

for i in range(40, 81, 10):
  intensity = i/100
  target_HR = (((220-age)-resting_HR)*intensity)+resting_HR
  print(f'운동 강도: {i}%일 때 목표 심박수: {target_HR} bpm')
```

## 3. while 문
* **특징**: 조건식이 '참(True)'인 동안 계속해서 문장을 반복 실행합니다. 횟수가 지정되지 않은 반복 상황에 적합합니다.
* **무한 루프**: 조건이 항상 참이면 프로그램이 끝나지 않고 계속 실행됩니다. 이를 방지하기 위해 반복문 내부에서 조건을 거짓으로 바꾸는 장치가 필요합니다.

```python
# 1~5까지 정수의 합
sum = 0
a = 1
while a <= 5:
  sum += a
  a += 1
  print(sum)
```

### break

반복문을 빠져 나올 때

```python
# 이름, 나이, 종료 입력 받아서 각각 출력
while True:
  answer = input('이름 or 나이 or 종료: ')
  if answer == '종료':
    print('종료하겠습니다.')
    break
  elif: answer.isdigit(): #isdigit() 문자열이 숫자형식이면 True, 그렇지 않으면 False를 반환
    print(f'{answer}세이군요')
  else:
    print(f'{answer}님, 환영합니다')
```

### 🎯 숫자 맞히기 게임
1. **컴퓨터가 생각한 수 맞히기**:
    * `random.randint(1, 100)`으로 정답을 생성합니다.
    * 사용자가 정답을 맞힐 때까지 `while` 문으로 입력을 반복합니다.
    * 입력값이 크면 "너무 높아요", 작으면 "너무 낮아요"를 출력하여 범위를 좁혀갑니다.
2. **사용자가 생각한 수 맞히기**:
    * 사용자가 숫자를 하나 정하면 컴퓨터가 `low`와 `high` 변수를 조정하며 정답을 찾아가는 로직입니다.

```python
# 1. 컴퓨터가 생각한 숫자 맞추기
#while문으로 반복하는데
#user가 입력한 숫자가 맞으면 반복문 종료.
#user가 입력한 숫자가 크면 크다고 출력
#user가 입력한 숫자가 작으면 작다고 출력.
#최종 도전 몇 회만에 정답을 맞췄는지 출력.

import random

print('1부터 100까지의 숫자 하나를 정했습니다.')
print('이 숫자를 맞춰보세요!')

answer = random.randint(1, 100)
time = 0 # 도전 횟수
user = int(input('숫자를 입력하세요: '))

while answer == user:
  time += 1
  if answer == user:
    print('정답을 맞췄습니다')
    print(f'도전 횟수: {time}')
    break
  elif answer > user:
    print('정답보다 작습니다.')
  else:
    print('정답보다 높습니다')

#선생님 코드

while True:
  user = int(input('1~100 사이의 숫자를 입력하세요'))
  time +=1
  if user == answer:
    break
  elif user > answer:
    print(f'{user}보다 작습니다')
  else:
    print(f'{user}보다 큽니다')

print(f'정답입니다. {time}회만에 맞췄습니다.')
```

```python
# 2. 사용자가 생각한 수 맞히기
print('1~100 사이 숫자 하나를 정하세요')
print('이 숫자를 맞혀보겠습니다!')

low = 1
high = 100
times = 0

num = (low+high) //2

#사용자가 컴퓨터가 제시한 숫자를 보고

answer = input(f'{num}인가요? ok / up / down') # 사용자가 답변을 해줌
times += 1
while True:
  num = (low+high) //2
  answer = input(f'{num}인가요? ok / up / down')
  times += 1
  if answer == 'ok':
    break
  elif answer == 'up':
    low = num + 1
  elif answer == 'down':
    high = num - 1
  else:
    print('잘못 입력했습니다')
```
