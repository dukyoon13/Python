## 1. 리스트의 개요
* **개념**: 리스트(list)는 여러 개의 데이터를 하나의 이름으로 묶어서 저장하는 자료형입니다.
* **필요성**: 역할이 비슷한 변수가 많아질 때 리스트를 사용하면 프로그램이 간결해지고 데이터 관리가 쉬워집니다.
* **리스트 생성**: 대괄호 `[]`를 사용하며, 항목은 쉼표(`,`)로 구분합니다.
    * 예: `prices = [1000, 2000, 3000]`
* **항목 접근 (Indexing)**: 리스트의 각 항목은 인덱스 번호를 가지며, **0번부터 시작**합니다.
    * 예: `prices[0]`은 첫 번째 항목인 1000을 의미합니다.

```python
prices = [1500,3200,2500,4000,5500]
for i in range(len(prices)): #len(prices): 0,1,2,3,4 5개의 인덱스
  prices[i] = prices[i] + prices[i]*0.05
print(prices)
```

```python
menus=['짜장면','짬뽕','볶음밥']
for i in range(len(menus)): #len(menus): 3개의 인덱스
  print(menus[i])
```

```python
# 빈 리스트 생성 후 값 추가하기
menus = []
menus.append('짜장면')
menus.append('짬뽕')
menus.append('볶음밥')
print(menus)
```

```python
# 1~10까지 정수를 세제곱한 수를 리스트 저장
cubes = []
for i in range(1,11):
  cubes.append(i**3)
print(cubes)
```

### 리스트 내포 (list comprehension):
[] 안에 `for문`과  `조건식`을 내포시켜 한줄로 생성


*   [조건식 for 변수 in 반복가능한것 if 조건문]



```python
cubes =[i**3 for i in range(1,11)]
print(cubes)
```

```python
#1~100 중에서 7의 배수를 리스트에 저장
sequence = []
for i in range(1,101):
  if i % 7 == 0:
    sequence.append(i)
print(sequence)
```

```python
sequence = [ i for i in range(1,101) if i % 7 == 0]
print(sequence)
```

```python
score =[71,80,60,90,65]
result = []

for i in score:
  if i >= 70:
    result.append("합격")
  else:
    result.append("불합격")
print(result)
```

```python
#리스트 내포
score =[71,80,60,90,65]
result = ["합격" if i >= 70 else "불합격" for i in score]
print(result)
```

## 2. 리스트 생성과 인덱싱

* **생성**: 대괄호(`[]`)를 사용하며, 각 요소는 쉼표(`,`)로 구분합니다.
    * 예: `fruits = ['apple', 'banana', 'cherry']`
* **인덱싱**: 리스트의 각 요소에 접근하기 위한 번호표로, **0번**부터 시작합니다.
    * `fruits[0]`은 'apple', `fruits[1]`은 'banana'를 가리킵니다.

```python
menus=['짜장면','짬뽕','볶음밥']
print(menus[1])
menus[2] = '탕수육' #인덱스 바꾸기
print(menus)

print(menus[-1]) # 인덱스 번호가 역으로 바뀜
print(menus[-2])
print(menus[-3])

for i in range(len(menus)): # 0 1 2
  print(menus[i])

```



*   len(리스트):리스트의 길이,항목의 전체 개수를 반환
*   enumerate():인덱스와 항목을 모두 반환



```python
fruits = ["apple", "banana", "cherry"]

for index, item in enumerate(fruits):
    print(index, item)
```

슬라이싱(slicing):리스트의 일부 항목 추출
리스트[시작 인덱스: 종료 인덱스]

```python
#슬라이싱
letters=['A','B','C','D','E',]
print(letters[1:3])
print(letters[:4])
print(letters[2:])
print(letters[1:-1])
```

## 3. 리스트 주요 메서드

리스트의 요소를 추가, 수정, 삭제하는 다양한 기능을 제공합니다.

* **추가**: `.append(값)` (끝에 추가), `.insert(위치, 값)` (특정 위치에 삽입), `.extend([리스트])` (다른 리스트와 합침)
* **삭제**: `.remove(값)` (첫 번째 일치하는 값 삭제), `.pop(위치)` (특정 위치의 값을 꺼내고 리스트에서 제거)

```python
pl = []
pl.append('파이썬')
pl.append('자바스크립트')
print(pl)
```

```python
pl = ['파이썬','자바스크립트']
pl.insert(1,'c언어')
print(pl)
```

```python
pl = ['파이썬','자바스크립트']
pl.extend(['c언어','c++'])
print(pl)
```

```python
pets = ['강아지','고양이','햄스터','너구리']
select1 = pets.pop(3) #인덱스 3번 추출되고 나머지는 그대로 저장
print(select1)
print(pets)

select2 = pets.pop() # 값 지정 안하면 맨 마지막 인덱스 삭제
print(select2)
print(pets)

del pets[0] # del키워드는 삭제한 항목은 반환되지 않음 (강아지 삭제)
print(pets)
```

```python
pets = ['강아지','고양이','햄스터','너구리','강아지']
pets.remove('강아지') #첫번째로 일치하는: 강아지 삭제
print(pets)

pets.clear() #리스트 전체 삭제
print(pets)
```

## 4. 리스트 탐색 및 정렬

* **탐색**: `len(리스트)` (길이 확인), `.count(값)` (특정 값의 개수 확인)
* **정렬**: `.sort()` (원본 리스트를 오름차순 정렬), `sorted(리스트)` (정렬된 새로운 리스트 반환)

```python
# sort() 원본을 직접 오름차순 정렬, 내림차순 sort(reverse=True)
numbers = [3,1,4,2]
numbers.sort()
print(numbers)
numbers.sort(reverse=True)
print(numbers)

#sorted() 원본은 유지, 오름차순 정렬된 새로운 리스트 반환,내림차순 sorted(리스트,reverse=True)
numbers_sorted= [5,8,7,6]
new_numbers = sorted(numbers_sorted)
print(numbers_sorted)
print(new_numbers)
new_numbers2 = sorted(numbers_sorted,reverse=True)
print(new_numbers2)
```

```python
#reverse() 리스트를 역순으로
nations = ['한국','미국','일본','중국']
nations.reverse()
print(nations)
```

```python
# 값 in 리스트 : 리스트에 값이 있으면 True,없으면 False
months = ['jan','feb','mar','apr','may','jun','jul','aug','sep','oct','nov','dec']
answer = input('월(영어)')

if answer in months:
  print(f'{months.index(answer)+1}월')
else:
  print('해당하는 월이 없습니다.')
#리스트.index(값) : 리스트에 값이 있으면 해당 항목의 인덱스 반환
```

```python
import random
colors = ['red','green','blue','yellow','purple']
#random.choice(리스트) - 리스트에서 무작위로 선택
print(random.choice(colors))

#random.sample(리스트,개수) - 리스트 개수만큼 중복 없이 무작위 선택
print(random.sample(colors,2))

#random.choices(리스트,k=개수) - 중복 허용
print(random.choices(colors,k=2))

#random.shuffle(리스트)
print(random.shuffle(colors))
```

## 삼각비 값 찾기 사전
리스트의 인덱싱 기능을 활용하여 특정 각도에 대한 삼각비 값을 출력하는 프로그램을 만들어 봅니다.
### 코드 로직
1. **데이터 저장**: 15개의 삼각비 명칭과 그에 대응하는 값을 리스트에 각각 저장합니다.
2. **사용자 입력**: 사용자로부터 0~14 사이의 인덱스 번호를 입력받습니다.
3. **결과 출력**: 입력받은 번호를 리스트의 인덱스로 사용하여 해당 각도와 값을 매칭해 보여줍니다.



```python
angles = ["sin0", "sin30", "sin45", "sin60", "sin90",
"cos0", "cos30", "cos45", "cos60", "cos90",
"tan0", "tan30", "tan45", "tan60", "tan90"]

values = ["0", "1/2", "√2/2", "√3/2", "1", # sin 0~90
"1", "√3/2", "√2/2", "1/2", "0",# cos 0~
"0", "√3/3", "1", "√3", "정의되지 않음" # tan 0~90
]
#zip 함수 이용 이름과 값을 짝지어 출력
for name, val in zip(angles,values):
  print(f'{name} = {val}')
```

```python


while True:
  B = input('삼각비를 입력하세요("종료" 작성시 종료됨): ')
  if B == '종료':
    print('종료합니다')
    break
  elif B in angles:
    print(f'{B} = {values[angles.index(B)]}')
  else:
    print('다시 입력하세요')
```
