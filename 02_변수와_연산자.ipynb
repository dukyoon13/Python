### f-문자열(f-string):

문자열 앞에 소문자 f 또는 대문자 F를 붙이고 문자열 안의 변수를 중괄호{}로 감싸서 표현하는 것.

* 변수 값이 출력되는 형식을 지정할 수 있음.
* 변수명 앞에 콜론(:)을 넣고 그 다음에 형식을 지정.
* 중괄호 안의 d는 정수를 의미하고, f는 부동소수점 수를 의미함.

```python
#print(f'문자열 {변수}')
name ='권덕윤'
age =20
print(f'이름은 {name}이고 나이는 {age}입니다')

print(f'{name:5}:{age:3}')
```

## 4. 연산자 (Operators)

산술 연산자를 통해 데이터를 계산합니다.

| 연산자 | 기능 | 예시 (a=10, b=3) | 결과 |
| --- | --- | --- | --- |
| `+` | 더하기 | `a + b` | 13 |
| `-` | 빼기 | `a - b` | 7 |
| `*` | 곱하기 | `a * b` | 30 |
| `/` | 나누기 (실수) | `a / b` | 3.333... |
| `//` | **몫** | `a // b` | 3 |
| `%` | **나머지** | `a % b` | 1 |
| `**` | 거듭제곱 | `a ** b` | 1000 |

```python
a = int(input('정수: '))
b = int(input('정수: '))
print(f'{a}+{b}={a+b}')
```

```python
print(f'{a} + {b} = {a+b}')
print(f'{a} - {b} = {a-b}')
print(f'{a} * {b} = {a*b}')
print(f'{a} / {b} = {a/b}')
print(f'{a} // {b} = {a//b}')
print(f'{a} % {b} = {a%b}')
print(f'{a} ** {b} = {a**b}') #a를 b번 거듭제곱.
```
