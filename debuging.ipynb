try: 실행

```python
nums = [1,2,3]
try:
    n = nums[5]
    print(n)
except:
    print("error 발생")
```

error  발생

```python
nums = []
for i in range(100):
    try:
        div = i - 50
        nums += [100/div]
    except Exception as e:
        print(e)
        print(i, div)
```

```python
nums = [1, 20, 300]
try:
    first = nums[0] / 0
except IndexError:
    print("인덱스 범위를 벗어났습니다")
except ZeroDivisionError:
    print("0으로 나눌 수 없습니다")

```
