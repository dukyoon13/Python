```python
#format()

num = 41.2
print('num = {}'.format(num))
print('num*2 = {}'.format(num*2))
print('numnum = {}'.format(str(num)*2))
```

```python
#upper(), lower()

word = 'Banana'
num_word = '41.2'
print(word.upper())
print(word.lower())
print(num_word.upper())
print(num_word.lower())
```

```python
#split(): 문자열을 일정 기준으로 나누어 리스트에 담음.

s = "It is Python!"
print(s.split()) #split()이라 하면, 공백문자를 기준으로 문자열을 나눔.

formula = 'x + y + z = 12'
left_right = formula.split('=')
left_vars = left_right[0].split('+') #left_right라는 리스트의 0번째(=첫 번째. ->0,1,2순) 요소를 가리킴(='x + y + z). 그것을 +를 기준으로 다시 나눔.
print(left_right)
print(left_vars)
```

```python
# append(): 리스트에 새로운 함수를 추가하는 함수.

nums = [3, 2, 1]
nums.append(0)
print(nums)
nums.append('number')
print(nums)
nums = [3, 2, 1] #재정의
nums.append([0, -1])
print(nums)
```

```python
#index(): 리스트 안에서 입력 요소의 위치를 찾는 함수.

words = ['orange', 'apple', 'grape']
idx = words.index('apple')
print(idx)
# =>1 : orange apple grape의 위치를 번호로 매기면 0, 1, 2다. 그러므로 apple의 위치를 인덱싱하면 1이 된다.

nums = [4, 3, 2, 1, 2]
i1 = nums.index(1)
i2 = nums.index(2)
print(i1)
print(i2)
```

```python
#pop(): 리스트 안에서 입력 숫자 인덱스의 요소를 결과로 내보내며 리스트 안에서 해당 위치의 요소를 제거함.
#pop은 꺼낸다는 의미.

nums = [4, 3, 2, 1]
last = nums.pop(-1) # pop(-1) 또는 pop()을 입력하면 리스트 내의 마지막 요소를 꺼낸다.
print(last, nums)
first = nums.pop(0)
print(first, nums)
```

```python
#remove()

nums = [1, 2, 3, 4, 3]
nums.remove(1)
print(nums)
nums.remove(3) # 리스트 내의 3을 모두 지우는 게 아니라, 3이라는 요소 단 하나를 지움. 3이 2개였다면 1개가 됨.
print(nums)

#nums.remove([0]) 실행되자 않음. 될 거라 생각했는데.
```
