## 1. 비교 연산자 (Comparison Operators)
* **개념**: 두 값의 크기를 비교하여 참(`True`) 또는 거짓(`False`)을 결과로 반환하는 연산자입니다.
* **주요 연산자**:

    * `>` / `<` : 크다 / 작다

````python
a, b, c = 10, 20, 30
print(f'{a} == {b}: {a==b}')
print(f'{a} != {b}: {a!=b}')
print(f'{a} > {b}: {a>b}')
print(f'{a} < {b}: {a<b}')
print(f'{a} >= {b}: {a>=b}')
print(f'{a} <= {b}: {a<=b}')
````

## 3. 논리 연산자 (Logical Operators)
* **개념**: 여러 개의 조건식을 조합하여 하나의 결과를 얻을 때 사용합니다.
* **종류**:
    * `and` : 두 조건이 모두 참일 때만 참
    * `or` : 두 조건 중 하나만 참이어도 참
    * `not` : 참을 거짓으로, 거짓을 참으로 반전

````python
#가위바위보 게임
import random

user = input('가위 / 바위 / 보: ')
com_rand = random.randint(1, 3)

if com_rand == 1:
  computer = '가위'
elif com_rand == 2:
  computer = '바위'
else:
  computer = '보'

print(f'나: {user}, 컴퓨터: {computer}')

# 승부 겨루기 : 무승부, 승, 패

if user == '가위':
  if computer == '보':
    print('user 승')
  elif computer == '바':
    print('computer 승')
  else:
    print('무승부')

if user == '바위':
  if computer == '가위':
    print('user 승')
  elif computer == '보':
    print('computer 승')
  else:
    print('무승부')
if user == '보':
  if computer == '바위':
    print('user 승')
  elif computer == '가위':
    print('computer 승')
  else:
    print('무승부')

#선생님 답안(승부 겨루기)

if user == computer:
  print('무승부')
elif user == '가위' and computer == '보':
  print('승')
elif user == '바위' and  computer == '가위':
  print('승')
elif user == '보' and computer == '바위':
  print('승')
else :
  print('패')
````

````python
#사각형 square, 삼각형 triangle, 원 circle 함수 3개


# 사각형은 가로 세로 입력 받아서 사각형 넓이는 가로*세로 값 반환

#삼각형은 밑변 base 높이 height 입력 받아서 삼각형 넓이는 밑*높이/2 값 반환

#원은 반지름 radius를 입력 받아서 math.pi*radius*radius 값 반환 (import math라는 라이브러리 사용. 불러와야함)

#도형의 종류를 입력 받아서, 변수  shape_type1, shape_type2
#사각형이면 사각형 함수 호출, 삼각형이면 삼각형 함수 호출, 원이면 원 함수 호출

#area 변수에는 각 도형의 넓이 값이 입력됨
#area 변수에 입력된 값을 비교해서 어떤 도형이 얼마만큼 더 넓은지 출력. 같을 경우에는 같다라고 출력.
````

### 📐 두 도형의 넓이 비교하기
1. **계산**: 사각형(가로×세로)과 삼각형(밑변×높이/2)의 정보를 각각 입력받아 넓이를 구합니다.
2. **비교**:
    * `if 넓이1 > 넓이2:` 첫 번째 도형이 더 넓다고 출력합니다.
    * `elif 넓이1 < 넓이2:` 두 번째 도형이 더 넓다고 출력합니다.
    * `else:` 두 도형의 넓이가 같다고 출력합니다.

````python
import math  #라이브러리 호출

#함수 선언
def square():
  width = int(input('사각형의 가로: '))
  height = int(input('사각형의 세로: '))
  return width * height
def triangle():
  base = int(input('삼각형의 밑변: '))
  height = int(input('삼각형의 높이: '))
  return base*height/2
def circle():
  radius = int(input('원의 반지름: '))
  return math.pi**radius

shape_type1 = input('도형 종류(사각형/삼각형/원): ')
shape_type2 = input('도형 종류(사각형/삼각형/원): ')

if shape_type1 == '사각형':
  area1 = square()
elif shape_type1 == '삼각형':
  area1 = triangle()
else:
  area1 = circle()

if shape_type2 == '사각형'
  area2 = square()
elif shape_type2 == '삼각형'
  area2 = triangle()
else:
  area2 = circle()

if area1 > area2:
  print(f'도형1(넓이: {area1})가 도형2(넓이: {area1})보다 {area1 - area2} 만큼 넓습니다')
elif area1 < area2:
  print(f'도형2(넓이: {area2})가 도형1(넓이: {area1})보다 {area2 - area1} 만큼 넓습니다')
else:
  print(f'두 도형의 넓이는 {area1}로 같습니다')
#shape_type1,2의 if문도 함수로 가능.
````
