```python
# 필요한 라이브러리 임포트
import pandas as pd
import numpy as np
from sklearn.datasets import load_iris
```

```python
# iris 데이터셋 불러오기
iris  = load_iris()
```

```python
iris
```

### Pandas dataFrame

```python
# 데이터 프레임을 변환
df = pd. DataFrame(iris.data, columns=['sepal_length', 'sepal_width', 'petal_length', 'petal_width'])
df['species'] = pd.Categorical.from_codes(iris.target, iris.target_names)
```

```python
#데이터 확인
print("=== 데이터 상위 5행 ===")
print(df.head())
```

```python
print("\n=== 칼럼 유형 확인 ===")
print(df.dtypes)
```

```python
print("\n=== 결측치 확인 ===")
print(df.isnull().sum())
```

```python
print("\n=== 종류별 데이터 개수 ===")
print(df['species'].value_continue())
```

### 평균 (Mean)


```python
# 단일 칼럼 평균
# print("꽃받침 길이 평균: {df['sepal_length].mean")
print(f'꽃받침 길이 평균: {df['sepal_length'].mean}')
```

```python
# 전체 숫자형 칼럼 평균
print("\n전체 칼럼 평균: ")
print(df.mean(numeric_only=True).round(2)) # 숫자형 칼럼만 계산하겠다.
```

```python
#종류별 평균 비교
print("\n 종류별 꽃잎 길이 평균: ")
print(df.groupby('species')['petal_length'].mean())
```

### 분위수 (Quantile / Percnetile)

```python
# 분위수 구하기
print("=== 분위수 (25%, 50%, 70%) ===")
print(df.quantile([0.25, 0.5, 0.75], numeric_only=True))
```

```python
# 특정 칼럼의 분위수
col = 'sepal_length'
Q1 = df[col].quantile(0.25)
Q2 = df[col].quantile(0.50)
Q3 = df[col].quantile(0.75)
IQR = Q3 - Q1
print(f"\n꽃받침 길이 분위수: ")
print(f" Q1(25%): {Q1}")
print(f" Q2(50%): {Q2}")
print(f" Q3(75%): {Q3}")
print(f" IQR    : {IQR}")
```

```python
  # 이상치 경계값
  lower = Q1 - 1.5 * IQR
  upper = Q3 + 1.5 * IQR
  print(f" 하한: {lower: .2f}")
  print(f" 상한: {upper: .2f}")
```

```python
#이상치 확인
outliers = df[(df[col] < lower) | (df[col] > upper)]
print(f"\n이상치 데이터 수: {len(outliers)}개")
#df[Ture, False, True] DataFrame에 불리언 시리즈를 전달하면 True로 표시된 행들만 필터링하여 새로운 DataFrame인 outliers를 반환
```

### 분산 (Variance)

```python
# 단일 컬럼 분산
print("꽃받침 길이 분산: ", df['sepal_length'].var())
print(f"꽃받침 길이 분산: {df['sepal_length'].var(): .2f}")
```

### 표준편차 (Standard Deviation)


```python
# 단일 칼럼 표준편차

#print("꽃받침 길이 표준편차: ", df['sepal_length'].std())
print(f"꽃받침 길이 표준편차: {df['sepal_length'].std(): .2f}")
```

```python
# 전체 칼럼 표준편차
print("\n === 전체 칼럼 표준편차 ===")
print(df.std(numeric_only=True).round(2))
```

```python
# 분산 vs 표준편차 비교 정리
print("\n === 분산 vs 표준편차 비교 ===")
summary = pd.DataFrame({
    '분산':df.var(numeric_only=True).round(2),
    '표준편차': df.std(numeric_only=True).round(4)
})
print(summary)
#표준편차(Standard Deviation): 분산의 제곱근으로,
#데이터가 평균으로부터 평균적으로 얼마나 떨어져 있는 지를 나타낸다.
#데이터의 원래 단위를 따르기 때문에 분산보다 직관적으로 이해하기 쉽다.

#petal_length(꽃잎 길이): 평균으로부터 가장 넓게 퍼져 있으며, 데이터 포인트들이 서로 크게 다르다는 것을 의미.
#sepal_width(꽃받침 너비): 데이터가 평균 주변에 가장 밀집되어 있으며, 데이터 포인트들이 서로 유사하는 것을 의미.
```

```python
# 평균 +- 표준편차 범위 (데이터의 약 68%가 이 범위에 속함)
col = 'sepal_length'
mean = df[col].mean()
std = df[col].std()
print(f"\n꽃받침 길이: 평균 +- 표준편차")
print(f" 평균: {mean: .2f}")
print(f" 표준편차: {std: .2f}")
print(f' 범위: {mean - std: .2f} ~ {mean + std: .2f}')
counting_in_range = df[(df[col] >= mean - std) & (df[col] <= mean + std)].shape[0]
```

```python
# 범주형 포함 전체 기술통계
print("\n === 번주형 포함 기술통계 ===")
print(df.dscribe(include= 'all').round(2))
```

```python
# 종류별 기술통계 (groupby + describe)
print("\n === 종류별 기술통계 ===")
print(df.groupby('species').describe())
```

```python
# 내가 원하는 통계만 선택해서 보기
print("\n=== 커스텀 기술통계 ===")
custom_stats = df.groupby('species')['petal_length'].agg([
    ('평균', 'mean'),
    ('중앙값', 'median'),
    ('표준편차', 'std'),
    ('최솟값', 'min'),
    ('최댓값', 'max')
]).round(3)
print(custom_stats)
```
